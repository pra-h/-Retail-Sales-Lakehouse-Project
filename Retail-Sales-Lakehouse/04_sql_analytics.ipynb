```python
%sql
SELECT 
    StockCode,
    Description,
    Total_Revenue,
    Total_Quantity_Sold,
    Unique_Customers
FROM retail_lakehouse.gold_product_performance
ORDER BY Total_Revenue DESC
LIMIT 10
```

```python
%sql
SELECT 
    Year,
    Month,
    Monthly_Revenue,
    Monthly_Orders,
    Monthly_Active_Customers,
    ROUND(Monthly_Revenue / Monthly_Orders, 2) AS Avg_Order_Value
FROM retail_lakehouse.gold_monthly_summary
ORDER BY Year, Month
```

```python
%sql
SELECT 
    Country,
    Total_Revenue,
    Total_Items_Sold,
    Total_Orders,
    Unique_Customers,
    ROUND(Total_Revenue / Total_Orders, 2) AS Revenue_Per_Order
FROM retail_lakehouse.gold_country_sales
ORDER BY Total_Revenue DESC
LIMIT 10
```

```python
%sql
SELECT 
    CustomerID,
    Total_Spent,
    Total_Orders,
    Total_Items_Bought,
    Avg_Order_Value
FROM retail_lakehouse.gold_customer_analysis
ORDER BY Total_Spent DESC
LIMIT 10
```

```python
%sql
SELECT 
    Year,
    Month,
    Day,
    Revenue,
    Total_Orders,
    Unique_Customers
FROM retail_lakehouse.gold_daily_sales
ORDER BY Revenue DESC
LIMIT 10
```

```python
%sql
SELECT 
    CASE 
        WHEN Total_Spent >= 5000 THEN 'VIP'
        WHEN Total_Spent >= 2000 THEN 'High Value'
        WHEN Total_Spent >= 500 THEN 'Medium Value'
        ELSE 'Low Value'
    END AS Customer_Segment,
    COUNT(*) AS Customer_Count,
    ROUND(SUM(Total_Spent), 2) AS Total_Revenue,
    ROUND(AVG(Total_Spent), 2) AS Avg_Spending,
    ROUND(AVG(Total_Orders), 0) AS Avg_Orders
FROM retail_lakehouse.gold_customer_analysis
GROUP BY 1
ORDER BY Avg_Spending DESC
```

```python
%sql
SELECT 
    CASE 
        WHEN Total_Revenue >= 50000 THEN 'Top Performer'
        WHEN Total_Revenue >= 10000 THEN 'Good Performer'
        WHEN Total_Revenue >= 1000 THEN 'Average'
        ELSE 'Low Performer'
    END AS Product_Category,
    COUNT(*) AS Product_Count,
    ROUND(SUM(Total_Revenue), 2) AS Category_Revenue,
    ROUND(AVG(Total_Revenue), 2) AS Avg_Product_Revenue
FROM retail_lakehouse.gold_product_performance
GROUP BY 1
ORDER BY Category_Revenue DESC
```

```python
%sql
WITH monthly_data AS (
    SELECT 
        Year,
        Month,
        Monthly_Revenue,
        LAG(Monthly_Revenue) OVER (ORDER BY Year, Month) AS Previous_Month_Revenue
    FROM retail_lakehouse.gold_monthly_summary
)
SELECT 
    Year,
    Month,
    Monthly_Revenue,
    Previous_Month_Revenue,
    ROUND(
        ((Monthly_Revenue - Previous_Month_Revenue) / Previous_Month_Revenue) * 100, 
        2
    ) AS Growth_Percentage
FROM monthly_data
ORDER BY Year, Month
```

```python
%sql
SELECT 
    Country,
    Total_Revenue,
    Unique_Customers,
    Total_Orders,
    ROUND(Total_Revenue / Unique_Customers, 2) AS Revenue_Per_Customer,
    ROUND(Total_Orders / Unique_Customers, 2) AS Orders_Per_Customer
FROM retail_lakehouse.gold_country_sales
WHERE Unique_Customers >= 5
ORDER BY Revenue_Per_Customer DESC
LIMIT 10
```

```python
%sql
SELECT 'Total Revenue' AS Metric, ROUND(SUM(Monthly_Revenue), 2) AS Value
FROM retail_lakehouse.gold_monthly_summary
UNION ALL
SELECT 'Avg Monthly Revenue', ROUND(AVG(Monthly_Revenue), 2)
FROM retail_lakehouse.gold_monthly_summary
UNION ALL
SELECT 'Total Customers', CAST(COUNT(*) AS STRING)
FROM retail_lakehouse.gold_customer_analysis
UNION ALL
SELECT 'Total Products', CAST(COUNT(*) AS STRING)
FROM retail_lakehouse.gold_product_performance
UNION ALL
SELECT 'Countries', CAST(COUNT(*) AS STRING)
FROM retail_lakehouse.gold_country_sales
UNION ALL
SELECT 'Total Orders', CAST(SUM(Monthly_Orders) AS STRING)
FROM retail_lakehouse.gold_monthly_summary
UNION ALL
SELECT 'Avg Order Value', ROUND(SUM(Monthly_Revenue) / SUM(Monthly_Orders), 2)
FROM retail_lakehouse.gold_monthly_summary
```
